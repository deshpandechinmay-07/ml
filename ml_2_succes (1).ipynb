```python
import pandas as pd
data=pd.read_excel("housing.xlsx")
data.head(10)
data["total_bedrooms"]=data["total_bedrooms"].fillna(data["total_bedrooms"].median())
data.isna().sum()
data.duplicated().sum()
data.head(10)
#data is cleaned
```

```python
data1=data.head(18000)
a=len(data)-18000
data2=data.head(a)
data1=data1.to_csv("house_train.csv")
data2=data2.to_csv("house_test.csv")
# data splited into train and test
```

```python
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd
X = data[[
    "longitude",
    "latitude",
    "housing_median_age",
    "total_rooms",
    "total_bedrooms",
    "population",
    "households",
    "median_income",
    "ocean_proximity"
]]

Y = data["median_house_value"]

X_encoded = pd.get_dummies(X)

model = RandomForestRegressor(random_state=42)

model.fit(X_encoded, Y)


# Test data
test_data = pd.read_csv("house_test.csv")

y_test = test_data["median_house_value"]

X_test = test_data[[
    "longitude",
    "latitude",
    "housing_median_age",
    "total_rooms",
    "total_bedrooms",
    "population",
    "households",
    "median_income",
    "ocean_proximity"
]]

# Encode categorical variable
X_test_encoded = pd.get_dummies(X_test)

# Make test columns identical to training columns
X_test_encoded = X_test_encoded.reindex(
    columns=X_encoded.columns,
    fill_value=0
)

# Prediction
predictions = model.predict(X_test_encoded)


# Evaluation
mae = mean_absolute_error(y_test, predictions)
mse = mean_squared_error(y_test, predictions)
rmse = np.sqrt(mse)
r2 = r2_score(y_test, predictions)

print(f"MAE:  {mae:.2f}")
print(f"RMSE: {rmse:.2f}")
print(f"R²:   {r2:.4f}")
```
