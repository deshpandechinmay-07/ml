```python
arr=[]
n=int(input("Enter the the total no students"))
for i in range(n):
  a = float(input(f"Enter the percentage of student{i+1}: "))
  if(a>0 and a<101):
    arr.append(a)
def selection(arr):
  for i in range(n):
    min_idx=i
    for j in range(i+1,n):
      if(arr[j]<arr[min_idx]):
        min_idx=j
    arr[i],arr[min_idx]=arr[min_idx],arr[i]
def bubble(arr):
  for i in range(n):
    for j in range(n-i-1):
      if(arr[j]>arr[j+1]):
        arr[j],arr[j+1]=arr[j+1],arr[j]
def top5(arr):
  for i in range(5):
    print(arr[i])
selection(arr)
print(f"After sorting{arr} by selection sort")
selection(arr)
print(f"After sorting{arr} by bubble sort")
bubble(arr)
print(f"The top 5 students are")
top5(arr)
```
